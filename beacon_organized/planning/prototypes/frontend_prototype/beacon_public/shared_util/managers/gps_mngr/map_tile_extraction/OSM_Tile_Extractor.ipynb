```python
# Easy-to-use extraction functions
# These functions use the imported modules for clean, simple extraction

# Quick extraction functions
async def extract_next_grid():
    """Extract the next available grid"""
    print("🎯 Extracting next grid...")
    result = await manager.extract_single()
    if result:
        print(f"✅ Success: {result[0]:,}/{result[1]:,} tiles extracted")
    return result

async def extract_multiple_grids_demo(num_grids=3):
    """Extract multiple grids with progress tracking"""
    print(f"🚀 Extracting {num_grids} grids...")
    result = await manager.extract_multiple(num_grids)
    print(f"🎉 Batch complete: {result[0]:,}/{result[1]:,} total tiles")
    return result

async def extract_coastal_grids_demo(num_grids=2):
    """Extract coastal grids for intersection variety"""
    print(f"🌊 Extracting {num_grids} coastal grids...")
    result = await manager.extract_coastal(num_grids)
    print(f"🏖️ Coastal extraction complete: {result[0]:,}/{result[1]:,} tiles")
    return result

# Monitoring functions
def check_progress():
    """Quick progress check"""
    manager.show_status()

def full_report():
    """Show comprehensive progress report"""
    manager.show_progress()

def show_upcoming_grids(count=5):
    """Show next grids to be extracted"""
    manager.show_next(count)

def show_state_breakdown():
    """Show progress by state"""
    manager.show_states()

def show_recent_work():
    """Show recent extraction activity"""
    manager.show_recent()

# Grid management functions
def initialize_fresh_grid():
    """Initialize a fresh grid system (WARNING: clears existing progress!)"""
    print("⚠️  WARNING: This will reset all progress!")
    confirm = input("Type 'YES' to confirm: ")
    if confirm == 'YES':
        return manager.initialize_grid_system()
    else:
        print("❌ Cancelled")
        return None

def get_grid_recommendations_demo(priority="coastal"):
    """Get recommended grids for extraction"""
    recs = manager.get_recommendations(priority)
    print(f"🎯 RECOMMENDED GRIDS ({priority}):")
    for i, grid_id in enumerate(recs[:10], 1):
        print(f"  {i:2}. {grid_id}")
    return recs

print("✅ Extraction control functions loaded!")
print("\n🎮 QUICK COMMANDS:")
print("   • await extract_next_grid() - Extract one grid")
print("   • await extract_multiple_grids_demo(n) - Extract n grids")
print("   • await extract_coastal_grids_demo(n) - Extract coastal grids")
print("   • check_progress() - Quick status")
print("   • full_report() - Complete report")
print("   • show_upcoming_grids(n) - Show next n grids")
print("   • get_grid_recommendations_demo() - Get recommendations")
```

```python
# US Shape Filter Functions
# Use these to verify and control the US-shaped grid display

def show_filtering_stats():
    """Show statistics about the filtering process"""
    # Get all grids
    with sqlite3.connect(manager.coverage_system.db_path) as conn:
        cursor = conn.execute('''
            SELECT status, COUNT(*) as count
            FROM grid_squares
            GROUP BY status
        ''')
        results = cursor.fetchall()
    
    print("📊 GRID FILTERING STATISTICS:")
    print("=" * 40)
    
    total_grids = 0
    for status, count in results:
        total_grids += count
        if status == 'pending':
            print(f"🔴 Pending (US land): {count:,} grids")
        elif status == 'completed':
            print(f"🟢 Completed: {count:,} grids")
        elif status == 'in_progress':
            print(f"🟡 In Progress: {count:,} grids")
        elif status == 'filtered_out':
            print(f"🗑️ Filtered (ocean/foreign): {count:,} grids")
        else:
            print(f"📋 {status}: {count:,} grids")
    
    pending_count = next((count for status, count in results if status == 'pending'), 0)
    filtered_count = next((count for status, count in results if status == 'filtered_out'), 0)
    
    if total_grids > 0:
        efficiency = (pending_count / total_grids) * 100
        print(f"\n📈 Efficiency: {efficiency:.1f}% of total area is useful US land")
        print(f"🌊 Ocean areas filtered: {filtered_count:,} grids")

def compare_maps():
    """Compare US-shaped vs full rectangular grid display"""
    import matplotlib.pyplot as plt
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))
    
    # Full rectangular grid
    viz1 = GridVisualization(manager.coverage_system)
    viz1.ax = ax1
    viz1.create_map(us_shape_only=False)
    ax1.set_title('🗺️ Full Rectangular Grid (All Areas)', fontsize=14)
    
    # US-shaped grid
    viz2 = GridVisualization(manager.coverage_system)
    viz2.ax = ax2
    viz2.create_map(us_shape_only=True)
    ax2.set_title('🇺🇸 US-Shaped Grid (Land Only)', fontsize=14)
    
    plt.tight_layout()
    plt.show()
    
    print("📊 Comparison complete!")
    print("Left: Shows all grids including ocean areas")
    print("Right: Shows only continental US land areas")

def reset_filtering():
    """Reset all grids back to pending (removes filtering)"""
    print("⚠️  WARNING: This will reset filtered grids back to pending!")
    confirm = input("Type 'YES' to confirm: ")
    
    if confirm == 'YES':
        with sqlite3.connect(manager.coverage_system.db_path) as conn:
            conn.execute('''
                UPDATE grid_squares
                SET status = 'pending'
                WHERE status = 'filtered_out'
            ''')
            conn.commit()
        
        print("✅ All grids reset to pending")
        print("🔄 Run update_grid_database_with_us_filter() to re-apply filtering")
    else:
        print("❌ Reset cancelled")

def apply_us_filter():
    """Apply US shape filtering to mark ocean areas as filtered"""
    filtered_count = update_grid_database_with_us_filter(manager.coverage_system)
    print(f"✅ Applied US shape filter - {filtered_count:,} ocean grids filtered out")

# Quick functions
def show_us_map():
    """Show US-shaped map"""
    show_grid_map(manager.coverage_system, us_shape_only=True)

def show_full_map():
    """Show full rectangular map"""
    show_grid_map(manager.coverage_system, us_shape_only=False)

print("✅ US Shape Filter functions loaded!")
print("\n🇺🇸 FILTER FUNCTIONS:")
print("   • show_filtering_stats() - View filtering statistics")
print("   • compare_maps() - Compare US-shaped vs full rectangular")
print("   • show_us_map() - Display US-shaped grid")
print("   • show_full_map() - Display full rectangular grid")
print("   • apply_us_filter() - Apply/reapply US shape filtering")
print("   • reset_filtering() - Reset all grids to pending")
```
